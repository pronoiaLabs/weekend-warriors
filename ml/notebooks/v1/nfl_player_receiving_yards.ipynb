## FEATURES grain

Player x game from `FEAT_PLAYER_GAME_ROLLING` + `FEAT_PLAYER_WEATHER`.
Train gate: `n_receiving_l5 > 0`. Labels come from CORE box scores, not the rolling table.

```python
%%sql -r player_grain
SELECT
    COUNT(*) AS rows,
    COUNT(IFF(r.is_completed AND r.season_type IN (2, 3), 1, NULL)) AS completed_rs_post,
    COUNT(IFF(r.is_completed AND r.season_type IN (2, 3) AND r.n_receiving_l5 > 0, 1, NULL)) AS gated
FROM NFL_PROD_DB.FEATURES.FEAT_PLAYER_GAME_ROLLING r
```

```python
%%sql -r eligible_by_season
SELECT r.season, COUNT(*) AS n
FROM NFL_PROD_DB.FEATURES.FEAT_PLAYER_GAME_ROLLING r
WHERE r.is_completed AND r.season_type IN (2, 3)
  AND r.n_receiving_l5 > 0
GROUP BY 1
ORDER BY 1
```

## Session and import

Kernel session via `get_active_session()`. Add `ml/` to `sys.path` so this
notebook can import the same module the laptop `uv run` uses.

```python
from pathlib import Path
import sys

here = Path.cwd().resolve()
for cand in (here, *here.parents):
    if (cand / "weekend_warriors_ml" / "pipeline.py").exists():
        if str(cand) not in sys.path:
            sys.path.insert(0, str(cand))
        break
else:
    raise FileNotFoundError(
        "weekend_warriors_ml not found. Put the repo ml/ folder in this Workspace."
    )

from snowflake.snowpark.context import get_active_session
from weekend_warriors_ml.specs import get_spec
from weekend_warriors_ml.pipeline import (
    cook,
    fit,
    inspect,
    log_experiment,
    register,
    score_batch,
    score_local,
)

SPEC = get_spec("NFL_PLAYER_RECEIVING_YARDS")
session = get_active_session()
print(SPEC.name, SPEC.task, len(SPEC.feature_columns), "features")
print(session.get_current_role(), session.get_current_warehouse())
```

```python
inspect(session, SPEC.name)
```

## Fit

HistGradientBoostingRegressor. Prints error vs a mean baseline.

```python
model, metrics = fit(session, SPEC.name)
metrics
```

## Register

Logs the experiment (best-effort), then `log_model` to `NFL_PROD_DB.ML` for SPCS.
`version_name` is whatever `log_model` auto-generates.

```python
try:
    log_experiment(session, metrics, SPEC.name)
except Exception as exc:
    print("Experiment tracking skipped:", exc)

version_name = register(session, model, SPEC.name)
version_name
```

## Score

Writes `NFL_PROD_DB.ML.PRED_PLAYER_RECEIVING_YARDS` from the in-kernel model (full slate).

```python
preds = score_local(session, model, version_name, SPEC.name)
preds.head(5)
```

```python
%%sql -r pred_summary
SELECT
    COUNT(*) AS rows,
    COUNT(label_receiving_yards) AS labeled,
    MIN(scored_at) AS scored_at,
    MIN(version_name) AS version_name
FROM NFL_PROD_DB.ML.PRED_PLAYER_RECEIVING_YARDS
```

```python
%%sql -r walkforward
SELECT
    COUNT(*) AS n_2025_rs_post,
    ROUND(AVG(ABS(pred_receiving_yards - label_receiving_yards)), 3) AS mae,
    ROUND(AVG(POW(pred_receiving_yards - label_receiving_yards, 2)), 3) AS mse
FROM NFL_PROD_DB.ML.PRED_PLAYER_RECEIVING_YARDS
WHERE season = 2025
  AND season_type IN (2, 3)
  AND label_receiving_yards IS NOT NULL
```
